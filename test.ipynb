```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_json("data/log/data_0.8_50.jsonl")

# get the response column (series)
data = df["response"]

# convert response dict in flat table (df)
data = pd.json_normalize(data)

# get the inference_tokens col (series)
inference_tokens = data["inference_tokens"]
durations = data["duration"]

# basic stats
stats = {
    "duration": {
        "min": float(durations.min() / 1_000_000_000),
        "max": float(durations.max() / 1_000_000_000),
        "avg": float(durations.mean() / 1_000_000_000),
    },
    "inference_tokens": {
        "min": int(inference_tokens.min()),
        "max": int(inference_tokens.max()),
        "avg": int(inference_tokens.mean()),
    },
}

# plot in time
plt.plot(inference_tokens.index, inference_tokens.values)
plt.show()

stats
```
